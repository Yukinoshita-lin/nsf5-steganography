```python
import numpy as np
from ns5_core import embed_string, extract_string, build_hamming, syndrome, permute_index

img = np.random.default_rng(0).integers(0, 256, (128, 128), dtype=np.uint8)
for method in ('matrix', 'nsF5'):
    stego, rep, nb = embed_string(img, 'wet paper demo', method=method, p=3)
    got = extract_string(stego, method=method, p=3)
    print(method, 'ok' if got == 'wet paper demo' else got,
          'changed', rep['cover_changed'])
```

```python
# 湿纸需要“危险像素”: 127/128/129 (|x-128|<=1)
danger = np.zeros_like(img)
danger[:, :] = img
danger[:20, :] = 128
stego, rep, nb = embed_string(danger, 'wet test 0123456789', method='nsF5', p=3)
print('nsF5 with danger zone ok:', extract_string(stego, method='nsF5', p=3) == 'wet test 0123456789')
```
