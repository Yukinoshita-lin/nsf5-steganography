```python
import image_io as IO
from ns5_core import embed_string, extract_string, get_image_hash

img = IO.load_as_gray('img/cover.png')
stego, report, _ = embed_string(img, 'secret', method='nsF5', p=3, password='A')
print('embed cover_hash:', report['cover_hash'][:16])
print('password A ok:', extract_string(stego, method='nsF5', p=3, password='A'))
```

```python
try:
    print(extract_string(stego, method='nsF5', p=3, password='B'))
except Exception as e:
    print('wrong password rejected:', type(e).__name__)
```

```python
tampered = stego.copy()
tampered[0, 0] ^= 1
print('hash changed:', get_image_hash(stego)[:16] != get_image_hash(tampered)[:16])
print('decode after tamper:', repr(extract_string(tampered, method='nsF5', p=3, password='A')))
```
