```python
import image_io as IO
from ns5_core import embed_string, extract_string
import steganalysis as SA

clean = IO.load_as_gray('img/cover.png')
stego, report, nbits = embed_string(clean, 'Hello nsF5!', method='nsF5', p=3)
print('changed pixels:', report['cover_changed'], '/ total', clean.size)
print('decoded:', extract_string(stego, method='nsF5', p=3))
```

```python
for name, im in (('clean', clean), ('stego', stego)):
    r = SA.analyze(im)
    print(name, 'Gn=%.3f' % r['RS_Gn'], 'chi2p=%.3f' % r['chi2_pvalue'],
          'prob=%.2f' % r['stego_probability'], r['verdict'])
```

```python
# 强化练习: 把消息加到接近满容量, 观察改动占比与检测概率如何上升
# (2026-09-14: 此处原写死 5000 字符, 超过 img/cover.png 的容量, 直接抛
#  ValueError —— notebook 从未被执行过, 所以坏了很久没人发现。)
from make_dataset import capacity_bytes
cap = capacity_bytes(clean.size, 3, head_pixels=0)   # p=3 时的可用字节数
cap_chars = int(cap * 0.9)                            # 留 10% 余量
print('img/cover.png 容量(p=3, 90%%): %d 字节' % cap_chars)
s2, rep2, nb2 = embed_string(clean, 'S' * cap_chars, method='nsF5', p=3)
r2 = SA.analyze(s2)
print('density=%.3f' % (rep2['cover_changed'] / clean.size),
      'prob=%.2f' % r2['stego_probability'], r2['verdict'])
# 顺带看看超容量时的报错是否清楚 (教学点: 容量是硬约束)
try:
    embed_string(clean, 'S' * (cap * 2), method='nsF5', p=3)
except ValueError as e:
    print('超容量 ->', str(e)[:80])
```
