```python
# 多轮统计: p=3 时平均改动次数应接近 0.875/块
changes = 0
for seed in range(400):
    rr = MD.demo_step(3, MD.random_block(3, seed=seed), '110')
    changes += int(rr['modified'] and rr['valid_col'])
print('avg changes per block: %.3f (theory %.3f)' % (changes / 400, (2**3 - 1) / 2**3))
```
