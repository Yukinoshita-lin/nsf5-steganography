```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 生成 128x128 渐变灰度图
y = np.linspace(0, 255, 128)[None, :]
x = np.linspace(0, 200, 128)[:, None]
img = np.clip(30 + x + y, 0, 255).astype(np.uint8)
Image.fromarray(img).save('output_demo.png')
print('shape:', img.shape, 'dtype:', img.dtype, 'range:', img.min(), img.max())
```

```python
v = 200
print(bin(v))            # 0b11001000
print('LSB =', v & 1)
print('flip ->', v ^ 1)
print('clear ->', v & 0b11111110)
```

```python
bitplane = (img >> 0) & 1   # LSB 位平面
print('LSB=1 比例: %.3f' % bitplane.mean())
fig, ax = plt.subplots(1, 2, figsize=(8, 4))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('original')
ax[1].imshow(bitplane * 255, cmap='gray'); ax[1].set_title('LSB plane')
plt.show()
```
